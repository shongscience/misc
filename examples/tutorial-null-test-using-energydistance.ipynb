### Null Hypothesis Test using Energy Distance 

> for multi-dimensionally featured data

```python
# === Required Imports ===

import numpy as np
from scipy.spatial.distance import cdist
import matplotlib.pyplot as plt  # Optional, for visualizing permutation results
```

#### === [0] Template Samples ===

```python
# Each row = 1 LAE, columns = [log10_Lyα_lum, REW, M_UV]
cosmos_sample = np.array([
    [42.1, 75, -19.5],
    [42.4, 110, -20.2],
    [42.0, 60, -18.9],
    [42.3, 90, -19.8],
    [42.5, 130, -20.5],
    [42.2, 85, -19.6],
    [42.1, 70, -19.2],
    [42.6, 150, -20.7],
    [42.3, 95, -19.9],
    [42.0, 65, -18.8]
])

tng_sample = np.array([
    [42.2, 80, -19.4],
    [42.5, 100, -20.1],
    [42.1, 65, -19.0],
    [42.4, 90, -19.7],
    [42.3, 105, -20.2],
    [42.0, 55, -18.7],
    [42.6, 140, -20.6],
    [42.5, 120, -20.3],
    [42.2, 75, -19.3],
    [42.3, 95, -20.0],
    [42.4, 110, -20.4],
    [42.1, 60, -18.9],
    [42.0, 50, -18.5],
    [42.5, 135, -20.5],
    [42.3, 100, -19.8]
])
```

#### === [1] Energy Distance Function ===

```python
from scipy.spatial.distance import cdist

def energy_distance(x, y):
    """
    Compute the energy distance between two samples x and y.

    Parameters:
        x (np.ndarray): Sample A, shape (n_x, n_features)
        y (np.ndarray): Sample B, shape (n_y, n_features)

    Returns:
        float: The energy distance between x and y
    """
    x = np.atleast_2d(x)
    y = np.atleast_2d(y)

    n_x = len(x)
    n_y = len(y)

    # Pairwise distances
    d_xx = cdist(x, x, metric='euclidean')
    d_yy = cdist(y, y, metric='euclidean')
    d_xy = cdist(x, y, metric='euclidean')

    # Energy distance formula (handles unequal sizes)
    e_dist = (
        2.0 * np.sum(d_xy) / (n_x * n_y)
        - np.sum(d_xx) / (n_x ** 2)
        - np.sum(d_yy) / (n_y ** 2)
    )
    return e_dist
```

```python
# Compute energy distance
ed = energy_distance(cosmos_sample, tng_sample)
print(f"Energy distance between triangle samples: {ed:.4f}")
```

#### === [2] Permutation Test ===

```python
def permutation_test_energy(x, y, n_perm=1000, seed=42):
    """
    Permutation test for the energy distance between two unequal-sized samples.

    Parameters:
        x (np.ndarray): Sample A, shape (n_x, n_features)
        y (np.ndarray): Sample B, shape (n_y, n_features)
        n_perm (int): Number of permutations to run
        seed (int): Random seed for reproducibility

    Returns:
        tuple: (observed_energy_distance, p_value, permuted_distances)
    """
    rng = np.random.default_rng(seed)
    observed = energy_distance(x, y)

    combined = np.vstack([x, y])
    n_x = len(x)
    n_total = len(combined)

    permuted_distances = []
    for _ in range(n_perm):
        indices = rng.permutation(n_total)
        x_perm = combined[indices[:n_x]]
        y_perm = combined[indices[n_x:]]
        e_dist_perm = energy_distance(x_perm, y_perm)
        permuted_distances.append(e_dist_perm)

    permuted_distances = np.array(permuted_distances)
    p_value = np.mean(permuted_distances >= observed)

    return observed, p_value, permuted_distances
```

```python
# Run the permutation test
ed_val, p_val, permuted_val = permutation_test_energy(cosmos_sample, tng_sample)
print(f"Energy Distance: {ed_val:.4f}, p-value: {p_val:.8f}")
```
